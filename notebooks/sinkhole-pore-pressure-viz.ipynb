# 間隙水圧計データの図化（学生向け）

## このノートブックでできること

- **data** フォルダに入れた計測CSVを読み、
- **params/base.yaml**（と必要なら **params/cases/〇〇.yaml**）の設定に従って、
- 間隙水圧・レーザー変位・動水勾配のグラフを作り、
- **output/（ケース名）/figures/** に画像として保存します。

図は画面には出さず、ファイルに保存するだけです。レポート用に `output` フォルダから図を開いて使ってください。

## 実行の流れ

1. 下の「設定」セルで **WORK_DIR** と **CASE_ID** を確認・変更（必要なら）。
2. そのあとの「実行」セルを上から順に実行（Shift+Enter など）。
3. 実行が終わったら、`output/（ケース名）/figures/` に図が保存されているか確認。

設定の意味は **params/base.yaml** と **params/README.md** に説明があります。

## 設定の意味（実行前に確認）

| 変数 | 意味 | よく使う値 |
|------|------|------------|
| **WORK_DIR** | データ・設定・出力を置く「作業フォルダ」のパス。 | `None`＝このノートがあるフォルダ（またはその親）を作業フォルダにする。データを別フォルダに置くときだけ `Path("…")` で指定。 |
| **CASE_ID** | どの実験（ケース）を図化するか。 | `None`＝data 内の全CSVをまとめて実行。`"sample"` のように文字列を指定すると、そのケースだけ実行（例：`"No7-incline-8mm"`）。 |

- **WORK_DIR** を変えたら、そのフォルダに **data**（CSV入り）と **params**（base.yaml 等）があるようにしてください。**output** は自動で作られます。
- **CASE_ID** は、data フォルダにある CSV の「ファイル名（拡張子なし）」と一致させます。例：`No7-incline-8mm.csv` → `CASE_ID = "No7-incline-8mm"`。

## 実行セル

下のセルを実行すると、**WORK_DIR** で指定した作業フォルダの **data** からCSVを読み、**params** の設定で図化し、**output** に保存します。進捗はセル実行後の出力で確認できます。

```python
# Phase 5: 1セルのみ。Python API で呼ぶ（CLI より PATH/sys.argv の影響を受けにくい）。
# ノート・モジュールがあるフォルダ と data/params/output があるフォルダ を分離して使う。
import sys
import os
from pathlib import Path

# -----------------------------------------------------------------------------
# 1) 作業フォルダ（data / params / output を置くフォルダ）
#    ここを指定すると、ノートやモジュールの場所と別フォルダでデータを扱える。
# -----------------------------------------------------------------------------
# ローカル: None ならカレントが作業フォルダ。別フォルダにしたいときだけ Path を指定。
WORK_DIR = None  # 例: Path("../sinkhole_analysis") または Path.home() / "MyProjects" / "間隙水圧_データ"

# -----------------------------------------------------------------------------
# 2) Google Colab かどうか
# -----------------------------------------------------------------------------
try:
    from google.colab import drive
    _is_colab = True
    drive.mount("/content/drive")
except ImportError:
    _is_colab = False

# -----------------------------------------------------------------------------
# 3) モジュール（poreviz）のパスを通す（ノート・コード側のフォルダ）
# -----------------------------------------------------------------------------
if _is_colab:
    _code_root = Path("/content/drive/MyDrive/Colab Notebooks/Pore-Water-viz")
    _work = Path("/content/drive/MyDrive/Colab Notebooks/sinkhole_analysis")
    _work.mkdir(parents=True, exist_ok=True)
    os.environ["POREVIZ_WORK"] = str(_work.resolve())
    _src = _code_root / "src"
    sys.path.insert(0, str(_src.resolve()))
else:
    # ローカル: ノートがある場所 or 1つ上をコードルートに
    _code_root = Path(".").resolve()
    if not (_code_root / "src").exists() and (_code_root.parent / "src").exists():
        _code_root = _code_root.parent
    _src = _code_root / "src"
    if str(_src) not in sys.path:
        sys.path.insert(0, str(_src))
    # 作業フォルダ: WORK_DIR を指定していればそこ、なければコードルート（data/params/output が同じ階層）
    os.environ["POREVIZ_WORK"] = str(Path(WORK_DIR).resolve() if WORK_DIR is not None else _code_root.resolve())

# パッケージが読めるか確認（追加した _src の直下に poreviz があるか）
if not (_src / "poreviz").exists():
    raise FileNotFoundError(f"poreviz パッケージが見つかりません: {_src} に poreviz がありません")

from poreviz import env, config, pipeline

# 実行環境判定（local/colab）と params/base.yaml によるパス解決
env_type = env.get_environment()
base_cfg, paths = config.load_base()
env.print_startup(paths)

# CASE_ID: None または "" で全件、指定で単体。進捗はセル出力で追える。
CASE_ID = None  # 例: "sample" で 1 件のみ

if CASE_ID:
    pipeline.run_case(CASE_ID, base_config=base_cfg, paths=paths)
else:
    pipeline.run_all(base_config=base_cfg, paths=paths)

# plt.show() は使わない。出力は output/{case_id}/ に保存のみ。
```
